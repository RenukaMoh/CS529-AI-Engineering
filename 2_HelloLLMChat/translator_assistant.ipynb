#### Demo - Language Translator 

```python
# Import the dotenv library
from dotenv import load_dotenv
# import os to access the environment variables
import os
# import the OpenAI library 
from openai import OpenAI
# Creates a web interface for the chatbot
import gradio as gr     
```

```python
# it's time to load the API keys into environment variables
load_dotenv()
```

```python
# Create a single global client (better than creating it inside the function)
client = OpenAI()
```

```python
# language_translator function

def language_translation(language: str, text: str) -> str:
    """
    Translate English text into the requested language using OpenAI Responses API.
    """
    # strip() - Removing leading whitespaces
    language = language.strip() 
    text = text.strip()
    
    # Checks empty string or None
    if not language:
        return "⚠️ Please enter a target language (e.g., Spanish, Hindi, French)."
    if not text:
        return "⚠️ Please enter some English text to translate."

    try:
        response = client.responses.create(
            model="gpt-5-mini",
            instructions=(
                "You are a professional language translator. "
                "Translate the user's text from English into the requested language. "
                "Respond ONLY with the translated text. Do not add explanations, quotes, or extra text."
            ),
            input=f"Translate the following English text into {language}: {text}"
        )

        # Convenience helper from the Responses API
        translated_text = response.output_text

        return translated_text.strip()

    except Exception as e:
        # Error message, add emojis Mac - Control + Command + Space, Windows - Windows Key + . (period)
        return f"❌ An error occurred: {e}"
```

#### Language Translator UI Screen View

```python
# Create a Gradio Block-based interface container with a title
demo = gr.Blocks()

with demo:
    # Display a heading, add emojis Mac - Control + Command + Space, Windows - Windows Key + . (period)
    gr.Markdown("## 🌐 Language Translator ")

    language_box = gr.Textbox(label="Target Language", placeholder="e.g., French, Hindi, Spanish")
    text_box = gr.Textbox(label="English Text", lines=4, placeholder="Enter English text here...")

    translate_btn = gr.Button("Translate")
    output_box = gr.Textbox(label="Translated Text", lines=4)

# Connect the button click to the language_translation function and define inputs/outputs
    translate_btn.click(language_translation, inputs=[language_box, text_box], outputs=output_box)

# Launch UI — automatically opens in browser
demo.launch(share=True, inbrowser=True, inline=False)
```
