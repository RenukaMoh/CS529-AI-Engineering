#### Read PDF content and display it
**Code Explanation**:

The following code reads text from a PDF file while hiding unnecessary warning messages. It begins by importing required libraries and uses redirect_stderr to suppress internal warnings from the pypdf library. Normally, some PDF files generate long warning messages that appear in the notebook output. redirect_stderr helps avoid this by sending all error and warning messages to a special invisible file called os.devnull, which simply throws them away. This keeps the output clean and easier to read.


Inside this protected block, the code loads a PDF file named aboutme.pdf using PdfReader. It then loops through each page, extracts text, and adds it to a variable called full_text. After processing all pages, it prints the combined extracted text. This allows you to view the full content of your PDF directly in the notebook without being distracted by warning messages.

- with → safely open a file and automatically close it
- "w" mode → open file for writing (previous content gets erased)
- Why used here? → to throw away error/warning messages by writing them into os.devnull


```python

# Provides access to system-level utilities
import sys
import os

# contextlib offers tools for managing temporary contexts.
# redirect_stderr is used here to temporarily hide warning/error messages
# so they don't appear in the notebook output.
from contextlib import redirect_stderr

# PdfReader allows us to open a PDF file and extract its text
from pypdf import PdfReader

# Suppress internal pypdf warnings
# Hides error messages by sending them to a special invisible file that throws them away- devnull is device
with open(os.devnull, "w") as f, redirect_stderr(f):
    reader = PdfReader("aboutme.pdf") # Upload your profile pdf here
    full_text = ""
    for page in reader.pages:
        text = page.extract_text()
        if text:
            full_text += text + "\n"

print(full_text)
```

```python
# Read the Summary Text File, upload your key strengths in the folder or oveeride the content
with open("keystrengths.txt", "r", encoding="utf-8") as f:
    key_strengths = f.read()

print(key_strengths)
```

```python
# Giving the name of the Profile owner
name = "Renuka Mohanraj" # Please provide your name here
```

```python
# Design the System Prompt
system_prompt = f"You are acting as {name}. You are answering questions on {name}'s profile details, \
particularly questions related to {name}'s career, background, skills, key strengths and experience. \
Your responsibility is to represent {name} for interactions on the website as faithfully as possible. \
You are given a full text of {name}'s profile and  which you can use to answer questions. \
Be professional and engaging, as if talking to a potential client or future employer who came across the profile. \
If you don't know the answer, say so."

system_prompt += f"\n\n## Summary:\n{full_text}\n\n## Key Strengths :\n{key_strengths}\n\n"
system_prompt += f"With this context, please chat with the user, always staying in character as {name}."
print(system_prompt)
```

```python
# Define a chat function, there are two roles in the chat: system and user
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
    return response.choices[0].message.content
```
