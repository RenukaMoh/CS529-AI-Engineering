### Creating First Agent
- The system will go with the default model set for agent model settings
- Can see the default model on Traces
- Refer the documentation https://openai.github.io/openai-agents-python/models/

```python

import agents

agent1 = Agent(
    name="Assistant",
    instructions="You are a helpful assistant",
)

result1 = await Runner.run(
    agent1,
    "Write a haiku about recursion in programming."
)

print(result1) # Return the Coroutine object
print("\nLLM Response for the prompt")
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
print(result1.final_output) # Return the respose
```

```python
# Make an agent with name, instructions, model, OPEN AI supports other model also
agent2 = Agent(
    name="Joke Generator", # Name of the agent
    instructions="You are a helpful assistant to tell funny jokes.", # Like System Prompt
    model="gpt-5-mini", # Model to use
)
print(agent2)
```

```python
# Use the agent to generate a joke with the help of the runner
result2 = await Runner.run(
    agent2,
    "Tell me a joke about Autonomous AI Agent."
)
print(result2.final_output)
```

```python
# Can see the traces from https://platform.openai.com/traces by logging into your account

with trace("Telling a joke"):
    result = await Runner.run(agent1, "Tell a joke about Autonomous AI Agents")
    print(result.final_output)
```

### Model Settings
Verbosity controls how much the model says in the final response.

- verbosity="low" → short, concise answer
- verbosity="medium" → balanced detail
- verbosity="high" → more explanation and 

Reasoning tells the model how much thinking effort to use before answering.
- low → quick, simple answer
- medium → balanced thinking
- high → more careful and detailed thinking
- max → deepest reasoning for complex tasks

```python
from openai.types.shared import Reasoning
from agents import Agent, ModelSettings

my_agent = Agent(
    name="My Agent",
    instructions="You're a helpful agent.",
    model="gpt-5.4",
    model_settings=ModelSettings(reasoning=Reasoning(effort="high"),
                                verbosity="low")
)
```

```python
# Use the agent to generate a joke with the help of the runner
my_result = await Runner.run(
    my_agent,
    "Tell me a joke about Autonomous AI Agent."
)
print(my_result.final_output)
```
