```python
from datetime import date, datetime

# Values for formatting
value = 12.3456
num = 1000000
rate = 0.256

today = date.today()
now = datetime.now()

# Preparing formatted outputs
two_decimals = f"{value:.2f}"
comma_format = f"{num:,}"
percent_format = f"{rate:.1%}"

date_dd_mm_yyyy = today.strftime("%d-%m-%Y")
date_month_name = today.strftime("%B %d, %Y")

time_24hr = now.strftime("%H:%M:%S")
time_12hr = now.strftime("%I:%M %p")

date_time = now.strftime("%d/%m/%Y %H:%M")

# Printing the table
print("\n Number, Date & Time Formatting Table\n")
print(f"{'Formatting Type':<30} {'Example Output':<30}")
print("-" * 60)

print(f"{'2 decimal places':<30} {two_decimals:<30}")
print(f"{'Comma separator':<30} {comma_format:<30}")
print(f"{'Percentage':<30} {percent_format:<30}")
print(f"{'Current date':<30} {str(today):<30}") # Need to convert date object to String
print(f"{'Date (DD-MM-YYYY)':<30} {date_dd_mm_yyyy:<30}") # <30 Left Align
print(f"{'Month name format':<30} {date_month_name:<30}")
print(f"{'Time (24-hour)':<30} {time_24hr:<30}")
print(f"{'Time (12-hour)':<30} {time_12hr:<30}")
print(f"{'Date + Time':<30} {date_time:<30}")
```
