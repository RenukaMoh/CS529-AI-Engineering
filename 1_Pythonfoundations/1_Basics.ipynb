```python
# Data Types
# Declaring Variables and Constants
PI = 3.14 # Constant, use UPPER CASE
count = 10
price = 99.5
is_active = True
result = None
# str is for both string and character
name = "Renuka"
grade = 'A' # No char type, single character
```

```python
# Printing the values
print("PI:", PI)
print("Count:", count)
print("Price:", price)
print("Name:", name)
print("Active:", is_active)
print("Result:", result)
print("Grade:", grade)
```

```python
# To print the type at runtime use type()
print(type(count))
print(type(price))
print(type(name))
print(type(is_active))
print(type(result))
print(type(grade))

```
