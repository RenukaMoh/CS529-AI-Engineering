```python
# Simple fuction no args, and return value
def greet(): 
  print("Hello!") 
 
```

```python
# Calling function
greet()
```

```python
# Passing argumets to the function
def welcome(name): 
    print("Welcome,", name) 

welcome("Renuka")
```

```python
# Passing multiple arguments and return value
def add(a,b): 
    return a + b 

result = add(4,10) 
print(result)
```

```python
# Function with default argument
def greet(name="Student"): 
    print("Hi", name) 

greet()
greet("Tom")
```

```python
#Function with multiple return values
def calc(a, b): 
    return a+b, a*b 

s, p = calc(3, 4) 
print(s, p)
```

```python
# Type hints help you show what type of data a variable, argument, or function returns.
def repeat(text: str, times: int) -> str: return text * times 

print(repeat("Hi", 3))
```
