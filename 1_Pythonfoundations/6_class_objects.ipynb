```python
# Person class demo: class, objects, attributes, methods

class Person:
    def __init__(self, name:str, age:int, city:str):
        # attributes
        self.name = name
        self.age = age
        self.city = city

    def introduce(self):
        print(f"Hi, I am {self.name}. I am {self.age} years old from {self.city}.")

    def move_to(self, new_city):
        self.city = new_city
        print(f"{self.name} moved to {self.city}.")

    def is_adult(self):
        return self.age >= 18


```

```python
# Create objects (instances)
p1 = Person("Bruce", 35, "Fairfield")
p2 = Person("Tom", 16, "Chicago")


# Use methods
p1.introduce()
p2.introduce()



# Change city for p2
p2.move_to("New York")

# Access attributes directly
print("p1 name:", p1.name)
print("p1 age:", p1.age)

print("p1 name:", p1.name)
print("p1 age:", p1.age)

# Check adult or not
print("Is Bruce an adult?", p1.is_adult())
print("Is Tom an adult?", p2.is_adult())
```

```python
from dataclasses import dataclass

@dataclass
class Car:
    brand: str
    model: str
    year: int
    price: float
```

```python
c1 = Car("Toyota", "Camry", 2020, 25000)
c2 = Car("Tesla", "Model 3", 2023, 42000)
```
