```python
print("=== 1. Simple if ===")
x = 10
if x > 5:
    print("x is greater than 5")
print()

# ----------------------------------------------------

print("=== 2. if-else ===")
x = 3
if x > 5:
    print("x is big")
else:
    print("x is small")
print()

# ----------------------------------------------------

print("=== 3. if-elif-else ===")
marks = 85
if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
else:
    print("Grade: C")
print()

# ----------------------------------------------------

print("=== 4. Nested if ===")
x = 10
y = 3
if x > 5:
    if y < 5:
        print("x > 5 AND y < 5 → Valid")
print()

# ----------------------------------------------------

print("=== 5. Logical operators (and/or/not) ===")
x = 8
if x > 5 and x < 10:
    print("x is in the range 6 to 9")

if x == 8 or x == 9:
    print("x is either 8 or 9")

flag = False
if not flag:
    print("flag is False, so print this")
print()

# ----------------------------------------------------

print("=== 6. Ternary (short if) ===")
age = 20
status = "Adult" if age >= 18 else "Minor"
print("Status:", status)
print()

# ----------------------------------------------------

print("=== 7. Multiple conditions in one line ===")
num = 15
if num > 10 and num < 20:
    print("num is between 11 and 19")
print()

# ----------------------------------------------------

print("=== All if statement demos completed! ===")
```

```python
print("=== 1. for loop (over a list) ===")
for x in [1, 2, 3]:
    print(x)
print()

# ----------------------------------------------------

print("=== 2. for loop with range(n) ===")
for i in range(3):
    print(i)
print()

# ----------------------------------------------------

print("=== 3. for loop with custom range(start, stop, step) ===")
for i in range(2, 10, 2):
    print(i)
print()

# ----------------------------------------------------

print("=== 4. while loop ===")
x = 1
while x <= 3:
    print(x)
    x += 1
print()

# ----------------------------------------------------

print("=== 5. break inside loop ===")
for x in range(5):
    if x == 3:
        break
    print(x)
print()

# ----------------------------------------------------

print("=== 6. continue inside loop ===")
for x in range(5):
    if x == 2:
        continue
    print(x)
print()

# ----------------------------------------------------

print("=== 7. Nested loops ===")
for i in range(2):
    for j in range(2):
        print(i, j)
print()

# ----------------------------------------------------

print("=== 8. Looping through a string ===")
for ch in "ABC":
    print(ch)
print()

# ----------------------------------------------------

print("=== 9. Looping with else ===")
for i in range(3):
    print(i)
else:
    print("Loop finished without break")
print()

# ----------------------------------------------------

print("=== All loop demos completed! ===")
```
