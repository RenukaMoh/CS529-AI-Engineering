### Agentic Pattern - Parallel Execution

```python
# Imports environment variables from a `.env` file.
from dotenv import load_dotenv

# Imports Agent, Runner, trace (for logging)
from agents import Agent, Runner, trace
import asyncio
# Print in the nice format
from IPython.display import Markdown, display
```

```python
load_dotenv()
```

```python
# Agent 1: Extract key points
keypoints_agent = Agent(
    name="KeyPointsAgent",
    instructions="Extract 3 key points from the given text."
)
```

```python
# Agent 2: Suggest improvements
improvement_agent = Agent(
    name="ImprovementAgent",
    instructions="Suggest 2 improvements for the given text."
)
```

```python
text = """
Artificial Intelligence (AI) has become a transformative force across many industries in recent years. 
In education, AI is being used to personalize learning experiences by adapting course content to 
individual student needs and learning speeds. Intelligent tutoring systems can provide instant 
feedback, helping students understand difficult concepts more effectively.

In the healthcare sector, AI assists doctors by analyzing medical images, predicting disease risks, 
and supporting clinical decision-making. These systems can process large volumes of patient data much 
faster than humans, leading to earlier diagnoses and improved treatment outcomes.

AI is also widely adopted in business and industry. Companies use AI-powered analytics to make 
data-driven decisions, optimize supply chains, and improve customer service through chatbots and 
virtual assistants. While AI offers significant benefits, it also raises concerns related to data 
privacy, ethical use, and the impact on employment. As AI continues to evolve, balancing innovation 
with responsible use remains a critical challenge.
"""
```

```python
# Defining a function to execute the parallel agents 
async def run_parallel_agents():
    # Run both agents in parallel
    keypoints_task = Runner.run(keypoints_agent, text)
    improvement_task = Runner.run(improvement_agent, text)
    keypoints_result, improvement_result = await asyncio.gather(
        keypoints_task,
        improvement_task
    )
    print("Key Points:")
    display(Markdown(keypoints_result.final_output))
   
    print("\nSuggested Improvements:")
    display(Markdown(improvement_result.final_output))
 
```

```python
# Call the function to Execute
await run_parallel_agents()
```
