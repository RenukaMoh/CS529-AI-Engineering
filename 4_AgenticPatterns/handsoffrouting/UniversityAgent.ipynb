### Agentic Pattern - Handoff and Routing
- Depends the user request teh responsibility handoff to the right agent such as Faculty Adviser, Nurse or Career Coach


```python
# Imports environment variables from a `.env` file.
from dotenv import load_dotenv

# Imports Agent, Runner, trace (for logging)
from agents import Agent, Runner, trace
```

```python

load_dotenv()
```

```python
# Use GPT-4o-mini model
useModel = "gpt-4o-mini"
```

```python
# Define the Faculty Advisor Agent
Faculty_Advisor_Agent = Agent(
    name="FacultyAdvisor",
    instructions="""You are Professor Alice. You help students with academic concerns like grades, course selections, study plans, or degree progress. 
    You are supportive and offer clear academic advice.""",
    model=useModel,
)
```

```python
# Define the Nurse Agent
Nurse_Agent = Agent(
    name="CampusNurse",
    instructions="""You are Nurse Lisa. You provide guidance on health, wellness, and emotional stress. 
    Your tone is warm, calm, and encouraging.""",
    model=useModel,
)
```

```python
# Define the Career Coach Agent
Career_Coach_Agent = Agent(
    name="CareerCoach",
    instructions="""You are Coach Tom. You help students plan their career path, internships, and job search strategy. 
    Your tone is motivating, practical, and forward-thinking.""",
    model=useModel,
)
```

#### In the context of AI agents:
A Triage Agent is a smart decision-making agent that:

Receives a user input (like a message or request)
Analyzes the content
Routes or delegates the task to the most appropriate specialized agent

```python
# Define the Triage Agent acting as a router/Main Agent
triage_agent = Agent(
    name="TriageAdvisor",
    instructions="""You are the university triage agent. A student will describe their concern.
Your job is to determine whether it should be routed to:
- FacultyAdvisor (for academics)
- CampusNurse (for health & stress)
- CareerCoach (for jobs & future career)

Be thoughtful and hand off to the most suitable agent.""",
    handoffs=[Faculty_Advisor_Agent, Nurse_Agent, Career_Coach_Agent],
    model=useModel,
)
```

```python
# Define the Student Agent as starting point like a user/human interaction
Student_Agent = Agent(
    name="Student",
    instructions="""You are an University level student writing a message to the support team.
    Use natural language to describe your concern or issue. Be honest and detailed.""",
    model=useModel,
)
```

```python
# Compose the student message
student_msg1 = "I feel very anxious lately and can't sleep."
student_msg2 = "I want to know the course schedule for the next semester."
student_msg3 = "I want to know who is my career coach."
```

```python
# Student agent sends the message
result = await Runner.run(
    Student_Agent,
    input=f"""You are a student at the university. Write a message explaining this issue: '{student_msg2}'"""
)
print("\nStudent Message:\n", result.final_output)

student_message = result.final_output
```

```python
# Tracing the message flow
with trace("Triage Agent"):
    result = await Runner.run(
    triage_agent,
    input=f"""The following message was received from a student. Please review it and route it to the appropriate support agent for handling:
{student_message}"""
)

print("\n Final Response:\n", result.final_output)
```
