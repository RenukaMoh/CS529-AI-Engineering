#### Demo - Hosted Tools(Pre-built-WebSearchTool and FileSearchTool)

```python
# Necessary imports
from dotenv import load_dotenv
from agents import Agent, Runner, trace,  WebSearchTool, FileSearchTool, CodeInterpreterTool
from agents.tool import CodeInterpreter
```

```python

load_dotenv()
```

```python
# Without Websearch Tool, can't get the latest information
agent = Agent(
    name="WebSearcher",
    instructions="You are a helpful assistant that answers questions using the web.",
    model='gpt-4o-mini'
)

with trace("WebSearcher"):
    result = await Runner.run(agent, "What is the current currency exchange rate for the Indian Rupee to the US Dollar?")

print(result.final_output)
```

**Note:** Try free websearch tools avialble and use it to avoid pricing

```python
# To get the latest information from the Web, need to use WebSearchTool()

agent = Agent(
    name="WebSearcher",
    instructions="You are a helpful assistant that answers questions using the web.",
    tools=[WebSearchTool()], # can use multiple tools inside the list []
    model='gpt-4o-mini'
)

with trace("WebSearcher"):
    result = await Runner.run(agent, "What is the current currency exchange rate for the Indian Rupee to the US Dollar?")

print(result.final_output)
```

### Using File Search Tool

```python
agent = Agent(
        name="File searcher",
        instructions="You are a helpful agent. Only use the given tool to answer the user prompts",
        tools=[
            FileSearchTool(
                vector_store_ids=["vs_688d106186f0819183e2acb776473873"], # Use your vector store id
            )
        ],
    )
```

```python
with trace("File search example"):
        result = await Runner.run(
            agent, "Be concise, and tell me Healthy Life Style Choices."
        )
        print(result.final_output)
```

### Using Code Interpreter Tool

```python
# Instantiate the tool
tool_config = CodeInterpreter(
    container={"type":"auto"},
    type="code_interpreter"       
)
codetool = CodeInterpreterTool(tool_config=tool_config)

# Create an agent
agent = Agent(
    name="CodeTool",
    instructions="You are an AI agent that writes and runs Python code to answer questions.",
    tools=[codetool]
)

result = await Runner.run(agent, "What is my monthly payment for a $800,000 mortgage at 6% for 30 years?")
print(result.final_output)
```
