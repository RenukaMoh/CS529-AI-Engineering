#### Demo - LLM as a Judge Pattern

In this demo, there are three teaching assistant agents: agent_detailed, agent_concise, and agent_engaging. The judge_agent reviews their answers to the student's question and scores them using a simple evaluation format.

1. Clarity (1-5)

2. Depth (1-5)

3. Usefulness to a university student (1-5)


```python
# import libraries
# Imports environment variables from a `.env` file.
from dotenv import load_dotenv

# Imports Agent, Runner, trace (for logging), and function_tool (for custom tools) from the agents module.
from agents import Agent, Runner, trace
```

```python
student_query = "Can you explain the difference between tools and handsoff in AI Agentic pattern with real world examples?"
```

```python

# Agent 1 - Detailed Explanation
agent_detailed = Agent(
    name="Detailed TA Agent",
    instructions="""
    You are an expert teaching assistant at a university. 
    You explain technical concepts deeply and clearly, using relevant academic and real-world examples for students.
    """,
    model="gpt-4o"
)

# Agent 2 - Concise Summary
agent_concise = Agent(
    name="Concise TA Agent",
    instructions="""
    You are a teaching assistant who gives short, clear, summary-style answers ideal for quick revision.
    """,
    model="gpt-3.5-turbo"
)

# Agent 3 - Engaging and Fun
agent_engaging = Agent(
    name="Engaging TA Agent",
    instructions="""
    You are a friendly teaching assistant who uses analogies, humor, and relatable examples to explain concepts in a fun and memorable way.
    """,
    model="gpt-4o-mini"
)
```

```python
# Get Responses from Each Agent
response_detailed = await Runner.run(agent_detailed,student_query)
response_concise = await Runner.run(agent_concise,student_query)
response_engaging = await Runner.run(agent_engaging,student_query)
```

```python
print("Response 1: \n", response_detailed.final_output)
print("Response 2: \n", response_concise.final_output)
print("Response 3: \n", response_engaging.final_output)
```

```python
instruction = """
    You are a neutral evaluator tasked with reviewing responses from three teaching assistant agents. 
    Your goal is to assess each response based on:
    1. Clarity (1-5)
    2. Depth (1-5)
    3. Usefulness to a university student (1-5)

    For each agent, provide:
    - A score breakdown
    - A brief comment on strengths and weaknesses

    Finally, declare which agent gave the best overall answer and why.
    Be fair, constructive, and evidence-based in your judgment.
    """
```

```python
# Judge Agent - Evaluates the three TA responses
judge_agent = Agent(
    name="TA Evaluation Judge",
    instructions=instruction,
    model="gpt-4-turbo"
)
```

```python
# Combine responses for evaluation
combined_responses = f"""
Student Question: {student_query}

Response from Detailed TA Agent:
{response_detailed}

Response from Concise TA Agent:
{response_concise}

Response from Engaging TA Agent:
{response_engaging}

As the Judge, please evaluate the three answers based on:
1. Clarity (1-5)
2. Depth (1-5)
3. Usefulness to a university student (1-5)

For each agent, provide a score breakdown and a brief comment.
Finally, declare which agent gave the best overall answer and give me the name of the agent
"""

```

```python

# Run Judge Agent
evaluation_result = await Runner.run(judge_agent, combined_responses)

# Display the result
print(evaluation_result.final_output)
```
