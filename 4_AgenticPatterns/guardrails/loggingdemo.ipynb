### Customer Support Agent with Logging - Scenario

```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def customer_support_agent(user_query):
    logging.info("Customer Support Agent started")
    logging.info(f"Received query: {user_query}")

    # Simple guardrail check
    if "weather" in user_query.lower():
        logging.warning("Input guardrail triggered for off-topic query")
        return "Sorry, I can only help with customer support questions."

    # Simulated tool/data lookup
    try:
        if "order" in user_query.lower():
            logging.info("Retrieving order details")
            return "Your order is being processed and will arrive in 3 days."
        else:
            logging.info("Providing general support response")
            return "How can I help you with your account or order?"
    except Exception as e:
        logging.error(f"Failed to process request: {e}")
        return "Sorry, something went wrong while handling your request."

```

```python
# Example run
response = customer_support_agent("Where is my order?")
print(response)
```

```python
# Example run
response = customer_support_agent("current weather in Iowa?")
print(response)
```
