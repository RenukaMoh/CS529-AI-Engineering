### Agentic Pattern - Deterministic Flow - ResumeBuilder

```python
# Imports environment variables from a `.env` file.
from dotenv import load_dotenv

# Imports Agent, Runner, trace (for logging)
from agents import Agent, Runner, trace

# Imports os for operating system operations.
import os
```

```python
load_dotenv()
```

```python
# Read the full content of profile.txt
with open('profile.txt', 'r') as file:  
    profile_content = file.read()
print(profile_content)
```

```python
# Build tailored instructions for the resume writer agent
def build_tailored_instructions(role: str, job_description: str, resume_format: str) -> str:
    return f"""
You are a professional resume writer. Based on the candidate's profile, create a well-structured resume tailored to the following job role:

Role: {role}

Job Description:
{job_description}

Format the resume using the following structure:
{resume_format}

Make the resume professional, relevant, and focused on matching the job requirements.
"""
```

```python
# Calling the function to build tailored instructions for the resume writer agent
role = "Online computer science adjunct professor positions"

job_description = """Faculty members are well known not just for their academic background, but also for their capacity to work closely and effectively with students. You'll find that they have a genuine interest in your success. 
The Adjunct Instructor of Computer Information Technology is responsible for the development and instruction of course materials in the fields of  programing, Artificial Intelligence and Machine Learning.
Teach online or in-person computer science (CSC) courses at the college, including preparing materials, supporting student learning, and recording and submitting grades.
"""
resume_format = """
- Title with the name and Role
- Contact Information
- Profile, Project Portfolio URLs
- Professional Summary
- Skills
- Education
- Research
- Projects
- Achievements
"""

writer_instructions = build_tailored_instructions(role, job_description, resume_format)
```

```python
print(writer_instructions)
```

```python
# Creating the first Resume Write Agent
ResumeWriterAgent = Agent(
    name="Resume Writer Agent",
    instructions=writer_instructions,
    model="gpt-4o-mini"
)
```

```python
# Instructions for the resume enhancer agent
enhancer_instructions = """
You are a professional resume enhancer. Take the draft resume provided and improve it by:

- Using strong, professional language and action verbs.
- Quantifying achievements where possible.
- Ensuring clarity, flow, and impact.
- Fixing any formatting or grammar issues.

Keep the original structure and sections intact.
"""
```

```python
# Creating the Second Resume Enhancer Agent
ResumeEnhancerAgent = Agent(
    name="Resume Enhancer Agent",
    instructions=enhancer_instructions,
    model="gpt-4o-mini"
)
```

```python
# Instructions for the resume reviewer agent
reviewer_instructions = """
You are a hiring manager reviewing a job applicant's resume. Your task is to:

- Review the enhanced resume for professionalism, clarity, and job fit.
- Make any final improvements if needed.
- Ensure it is ready for submission.

Return the final version of the resume. Do not add explanations or feedback—just the polished resume text.
"""
```

```python
# Creating the Third Resume Reviewer Agent
ResumeReviewerAgent = Agent(
    name="Resume Reviewer Agent",
    instructions=reviewer_instructions,
    model="gpt-4o-mini"
)
```

```python
# Running the Resume Writer Agent to generate a tailored resume by passing the profile content
writer_result = await Runner.run(ResumeWriterAgent, profile_content)
tailored_resume = writer_result.final_output
print(tailored_resume)
```

```python
# Running the Resume Enhancer Agent to enhance the resume by passing the tailored resume
enhancer_result = await Runner.run(ResumeEnhancerAgent, tailored_resume)
enhanced_resume = enhancer_result.final_output
```

```python
# Running the Resume Reviewer Agent to review the enhanced resume
reviewer_result = await Runner.run(ResumeReviewerAgent, enhanced_resume)
final_resume = reviewer_result.final_output
```

```python
# Print the final polished resume
print(final_resume)
```

```python
# Write to a .txt file
with open("mycv.txt", "w") as file:
    file.write(final_resume)

print("String successfully written to output.txt")
```
